```python
import pandas as pd
```

```python
train_data = pd.read_csv('train.csv')
test_data = pd.read_csv('test.csv')
```

```python
train_data.head()
```

```python
train_data = train_data.drop(['PassengerId','Name','Cabin'],axis='columns')
test_data = test_data.drop(['PassengerId','Name','Cabin'],axis='columns')
```

```python
train_data.isnull().sum()
```

```python
train_data = train_data.dropna()
test_data = test_data.dropna()
```

```python
print(train_data.isnull().sum())
print(test_data.isnull().sum())
```

```python
from sklearn.preprocessing import OrdinalEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
train_data.head()
```

```python
oe = OrdinalEncoder()
```

```python
oe.fit(train_data[['HomePlanet','CryoSleep','Destination','VIP','Transported']])
oe.fit(test_data[['HomePlanet','CryoSleep','Destination','VIP']])
```

```python
train_data[['HomePlanet','CryoSleep','Destination','VIP','Transported']] = oe.transform(train_data
                                                                                               [['HomePlanet','CryoSleep',
                                                                                                 'Destination','VIP','Transported']])
```
